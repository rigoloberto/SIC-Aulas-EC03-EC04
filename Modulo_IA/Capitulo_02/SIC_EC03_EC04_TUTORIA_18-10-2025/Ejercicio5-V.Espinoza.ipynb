```python
import sympy as sp

A = sp.Matrix([[1, 2], [3, 4]])
B = sp.Matrix([[0, 2], [1, 1]])

#1. operaciones basicas
Amasb = A + B
Amenosb = A - B
Amultb = A * B
A_traspuesta = A.T
detB = B.det()
inversaB = B.inv() if B.det() != 0 else None

print("\n")
print("Operaciones basicas:")
print("A + B =", Amasb)
print("A - B =", Amenosb)
print("A x B =", Amultb)
print("Transpuesta de A =", A_traspuesta)
print("Determinante de B =", detB)
print("Inversa de B =", inversaB)
print("\n")

#2. vectores
u = sp.Matrix([2, 3])
v = sp.Matrix([1, -1])

#       p.interno y ortogonalidad
dot_uv = (u.T * v)[0]
ortogonal = dot_uv == 0
#       magnitudes
mg_u = sp.sqrt((u.T * u)[0])
mg_v = sp.sqrt((v.T * v)[0])
#       angulo entre vectores
cos_theta = dot_uv / (mg_u * mg_v)
theta = sp.acos(cos_theta) * 180 / sp.pi
print("Vectores:")
print("u·v =", dot_uv)
print("¿Ortogonales?", ortogonal)
print("|u| =", mg_u)
print("|v| =", mg_v)
print("Ángulo (°) =", sp.N(theta))
print("\n")
#3. Derivadas
x, y = sp.symbols('x y')
g = x**2 * y + sp.sin(x*y)
dg_dx = sp.diff(g, x)
dg_dy = sp.diff(g, y)
print("Derivadas:")
print("g(x,y) =", g)
print("∂g/∂x  =", dg_dx)
print("∂g/∂y  =", dg_dy)
print("\n")
```
